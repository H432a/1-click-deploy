```python
import numpy as np
import matplotlib.pyplot as plt

data = np.loadtxt("/content/sample_data/regression (1).csv", delimiter=",",skiprows=1)

X=data[:,0]
Y=data[:,1]
n=len(X)

print("Mean Experience:", np.mean(X))
print("Mean Salary:", np.mean(Y))
print("Min Salary:", np.min(Y))
print("Max Salary:", np.max(Y))

var_X = np.sum((X - np.mean(X)) ** 2) / len(X)
var_Y = np.sum((Y - np.mean(Y)) ** 2) / len(Y)

print("Variance of Experience:", round(var_X, 4))
print("Variance of Salary:", round(var_Y, 2))
```

```python
def mean_squared_error(y_true, y_pred):
    return (1 / len(y_true)) * np.sum((y_true - y_pred) ** 2)


w = 0.0          # weight
b = 0.0          # bias
learning_rate = 0.01
epochs = 1000

losses = []


for epoch in range(epochs):

    # Prediction
    Y_pred = w * X + b

    # Compute loss
    loss = mean_squared_error(Y, Y_pred)
    losses.append(loss)

    # Compute gradients
    dw = (-2 / n) * np.sum(X * (Y - Y_pred))
    db = (-2 / n) * np.sum(Y - Y_pred)

    # Update parameters
    w = w - learning_rate * dw
    b = b - learning_rate * db

    # Optional: print loss every 100 epochs
    if epoch % 100 == 0:
        print(f"Epoch {epoch}, Loss: {loss}")

# Final parameters
print("\nTraining Completed")
print("Final Weight (w):", w)
print("Final Bias (b):", b)
```

```python
Y_pred = w * X + b

# Compare first 5 values
print("Actual vs Predicted (First 5):")
for i in range(5):
    print(f"Actual: {Y[i]:.2f}, Predicted: {Y_pred[i]:.2f}")

MSE = (1 / n) * np.sum((Y - Y_pred) ** 2)
MAE = (1 / n) * np.sum(np.abs(Y - Y_pred))

print("\nEvaluation Metrics")
print("Mean Squared Error (MSE):", MSE)
print("Mean Absolute Error (MAE):", MAE)



# Actual vs Predicted
plt.figure(figsize=(8,5))
plt.scatter(X, Y, color='blue', label='Actual Salary')
plt.plot(X, Y_pred, color='red', label='Predicted Salary')
plt.xlabel("Years of Experience")
plt.ylabel("Salary")
plt.title("Actual vs Predicted Salary")
plt.legend()
plt.show()

```

```python
plt.plot(losses)
plt.yscale('log')   # Log scale on Y-axis
plt.xlabel("Epochs")
plt.ylabel("MSE Loss (log scale)")
plt.title("Loss vs Epochs (Log Scale)")
plt.show()
```
